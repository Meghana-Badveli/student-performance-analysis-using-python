```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
student=pd.read_csv('student_data.csv')
```

```python
student = student.drop(columns=['Unnamed: 0','LunchType','IsFirstChild','NrSiblings','TransportMeans'],errors='ignore')
student
student.isnull().sum()
categorical_cols = ['Gender','EthnicGroup','ParentEduc','TestPrep','ParentMaritalStatus','PracticeSport','WklyStudyHours']
for col in categorical_cols:
     student[col] = student[col].fillna(student[col].mode()[0])
student.isnull().sum()
student.drop_duplicates(inplace=True)
student
score_cols = ['MathScore', 'ReadingScore', 'WritingScore']
student['OverallScore'] = student[score_cols].mean(axis=1)
student
```

```python
plt.figure(figsize=(14, 10), facecolor='cyan')
plt.suptitle('Dashboard-1: Student Performance', fontsize=24,color='purple', fontweight='bold')

plt.subplot(2, 2, 1)
plt.title('Parent Education Degrees', fontsize=16, fontweight='bold')
p = student['ParentEduc'].value_counts()
plt.pie(p, labels=p.index,colors=['lightgreen', 'orange', 'pink', 'purple', 'violet', '#DA6B64'],autopct='%1.1f%%', startangle=90)

plt.subplot(2, 2, 2)
plt.title('Gender Count', fontsize=16, fontweight='bold')
count = sns.countplot(x='Gender', data=student, color='#DA6B64')
for i in count.containers:
    count.bar_label(i, fmt='%d', padding=-135)

plt.subplot(2, 2, 3)
plt.title('Overall Score by Gender', fontsize=16, fontweight='bold')
student['OverallScore'] = student[['MathScore', 'ReadingScore', 'WritingScore']].mean(axis=1)
count = sns.barplot(x='Gender', y='OverallScore', data=student,color='yellow', errorbar=None)
for i in count.containers:
    count.bar_label(i, fmt='%.2f', padding=-135)
plt.xlabel('Gender')
plt.ylabel('Average Overall Score')

plt.subplot(2, 2, 4)
count = sns.barplot(x='WklyStudyHours', y='OverallScore',hue='Gender', data=student,palette='Set1', errorbar=None)
for i in count.containers:
    count.bar_label(i, fmt='%.1f', padding=-135, fontsize=8)
plt.title('Study Hours vs Overall Score by Gender',fontsize=16,fontweight='bold')
plt.xlabel('Weekly Study Hours')
plt.ylabel('Average Overall Score')
plt.xticks(rotation=15)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
plt.figure(figsize=(18, 11), facecolor='lavender')
plt.suptitle('Dashboard-2: Factors Associated with Student Scores', fontsize=25, color='purple', fontweight='bold')

# 1. Gender Ratio (Donut Chart) - Coral & Sage Green
plt.subplot(2, 3, 1)
g = student['Gender'].value_counts()
plt.pie(g, labels=g.index, autopct='%1.1f%%', wedgeprops={'width': 0.6}, startangle=135, pctdistance=0.68, colors=['#E07A5F', '#81B29A'])
plt.title('Gender Ratio', fontweight='bold', fontsize=14)

# 2. Gender vs All Scores (Horizontal Bar) - Teal, Slate, Muted Red
plt.subplot(2, 3, 2)
p = student.pivot_table(values=['MathScore', 'ReadingScore', 'WritingScore'], index='Gender', aggfunc='mean')
ax = p.plot(kind='barh', ax=plt.gca(), color=['#3D5A80', '#98C1D9', '#EE6C4D'])
for i in ax.containers:
    ax.bar_label(i, fmt='%.1f', padding=3, fontsize=8)
plt.title('Gender vs Math, Reading and Writing', fontweight='bold', fontsize=14)
plt.xlabel('Average Score')
plt.ylabel('Gender')
plt.xticks(rotation=0)

# 3. Gender vs Pass/Fail (Vertical Bar) - Dark Purple & Gold
plt.subplot(2, 3, 3)
student['OverallScore'] = student[['MathScore', 'ReadingScore', 'WritingScore']].mean(axis=1)
student['Result'] = student['OverallScore'].ge(50).map({True: 'Pass', False: 'Fail'})
p = student.pivot_table(values='OverallScore', index='Gender', columns='Result', aggfunc='count', fill_value=0)
ax = p.plot(kind='bar', ax=plt.gca(), color=['#4A4E69', '#F2CC8F'])
for i in ax.containers:
    ax.bar_label(i, fmt='%d', padding=4, fontsize=8)
plt.title('Gender vs Pass/Fail Count', fontweight='bold', fontsize=14)
plt.xlabel('Gender')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)

# 4. Parent Education and Gender vs All Scores (Heatmap) - Viridis
plt.subplot(2, 3, 4)
p = student.pivot_table(values=['MathScore', 'ReadingScore', 'WritingScore'], index=['ParentEduc', 'Gender'], aggfunc='mean')
sns.heatmap(p, annot=True, cmap='viridis', fmt='.2f')
plt.title('Parent Education & Gender vs Scores', fontweight='bold', fontsize=12)

# 5. Gender and Test Preparation vs All Scores (Heatmap) - YlOrRd
plt.subplot(2, 3, 5)
p = student.pivot_table(values=['MathScore', 'ReadingScore', 'WritingScore'], index=['Gender', 'TestPrep'], aggfunc='mean')
sns.heatmap(p, annot=True, cmap='YlOrRd', fmt='.2f')
plt.title('Gender & Test Prep vs Scores', fontweight='bold', fontsize=12)

# 6. Gender vs All Three Scores by Sports Practice (Heatmap) - Mako
plt.subplot(2, 3, 6)
p = student.pivot_table(values=['MathScore', 'ReadingScore', 'WritingScore'], index=['Gender', 'PracticeSport'], aggfunc='mean')
sns.heatmap(p, annot=True, cmap='mako', fmt='.1f')
plt.title('Gender & Sports Practice vs Scores', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.show()
```
